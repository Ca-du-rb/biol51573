### Printing quotation marks

##### Write two print statements that give the following output:

She said, 'hello world'

She said, "hello world"


```python
print("She said, 'hello world'")
print('She said, "hello world"')
print("She said, \"hello world\"") # Escaping the special character takes away its operability, turning it into a common character (literal interpretation by the compiler)
```

### Store and print a variable

```python
# define some strings
first = 'bob'
last = ' dylan'

# concatenate those strings
full = first + last

# print the strings
print(first)
print(last)
print(full)
print(first + last + "first")
print(first + ' ' + last + ' is the queen')
```

### Get the length of a string

```python
# the long (but readable) way
name_length = len(full)
print(name_length)

# a more concise way to do it, functions within functions
print(len(full))
```

### Print strings and numbers

```python
# print strings and integers
print("There are " + str(name_length - 1) + " characters in Bob Dylan's name.")
print("There are", str(name_length - 1), "characters in Bob Dylan's name.")

# print strings

# formatted print
print(f"There are {name_length - 1} characters in Bob Dylan's name.")
```

## Some operations on strings

```python
# methods to change the case of a string
print(first, last)
print(first.upper())
print(last.lower())

# type/class of variable
print(type(first))

# make the name properly cased 
db_name = 'AnDy'

# lower-case everything then capitalize so only the first letter of the string is upper-case
print(db_name.lower().capitalize())

# a literal string can be given as input
print("andy".capitalize())
```

### Problem #1: Write a program that asks for the user's name and age, then prints a personalized greeting using string formatting (`.format`).

For *Andy* and *32*, the output would be:

Hello, Andy! You are 32 years old.


```python
# code here
name = input("Please provide your name: ")
age = input("Please provide your age: ")

print(f"Hello, {name.lower().capitalize()}! You are {age} years old.")
print("Hello, {}! You are {} years old.".format(name.lower().capitalize(), age))
```

### Problem #2: Ask the user for a big number. If the user enters 1000000 (one million), the output will be:

This sure is a big number: 1,000,000

This sure is a big number: 1.000000e+06

```python
big_num = input('Enter your big number here:')

print(type(big_num))

print(f"This sure is a big number: {int(big_num):,d}") # digits notation

print(f"This sure is a big number: {int(big_num): e}") # exponential notation
```

### Problem #4: Tip Calculator

Ask the user for the bill amount (float) and tip percentage (integer). Calculate the tip and total, then print them neatly.

Enter the bill amount: `42.75`

Enter the tip percentage: `18`

Bill: `$42.75`

Tip (18%): `$7.70`

\--------------------

Total: `$50.45`


```python
# problem 5 solution
bill = float(input("Enter the bill amount: "))

tip = int(input("Enter the tip percentage: "))

tipQTD = tip/100*bill

total =  bill + tipQTD

print(f"Bill: ${round(bill, 2)}")

print(f"Tip ({tip}%): ${round(tipQTD, 2)}")

print("-------------------")

print(f"Total: ${round(total, 2)}")


#print(f"Bill: ${round(bill, 2)}\nTip ({tip}%): ${round(tipQTD, 2)}\n-------------------\nTotal: ${round(total, 2)}")
```
