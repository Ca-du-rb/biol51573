```python
from collections import defaultdict
import csv

# declare some dictionaries

# dictionary with string as key, but the type declared in the argument must be the value
presidents = defaultdict(str) # key = president number, value = president name
party_count = defaultdict(int) # key = party, value = count of presidents in that party
# without defaultdict, we'd declare a dictionary this way: parties = {}

# store the path and open the presidents.csv file
input_file = "../data/presidents.csv"

# read the file line by line
# president is interpreted as each individual line in the open file

with open(input_file, "r") as presfile:
    # create a csv reader object to parse the file
    reader = csv.reader(presfile, delimiter=",")

    # skip the header line
    header = next(reader)

    # read the file line by line
    for line in reader: 

        # print just to make sure we are reading the file okay
        print(line)

        # strip the line breaks <- deprecated
        # line = line.strip() # line is now a list, not a string to be stripped

        # split each line into a list <- deprecated and being replaced with csv reader
        # columns = president.split(',')

        # did we split it? yes, we did (prints first element of the list that line is now represented by)
        print(line[0])

        # load our presidents {} dictionary, with num as e key, and name as the main object
        num = line[0].strip()
        name = line[1].strip()
        presidents[num] = name
        # or
        # presidents[columns[0]] = columns[1]

        # load the party_count dictionary
        party = line[5].strip()
        party_count[party] += 1
```

```python
# instead of printing, let's write our output to a file
outfile = open("pres.out", "w")

# another way to loop over a dictionary
for num, name in presidents.items():
    outfile.write(f"president number {num} was {name}")

# data doesn't get written until you close it
outfile.close()
```

```python
# loop over party_count dictionary
for party, count in party_count.items():
    print(f'{party}: {count}')
```
