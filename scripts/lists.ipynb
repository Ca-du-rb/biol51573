## Working with lists

```python
# declare an empty string
bob = ''

# declare a list
beatles = []
```

### What kind of variable is this?

```python
print(type(bob))
print(type(beatles))    
```

### Populate and print a list

```python
beatles = ["John", "Paul", "George", "Ringo", "Stuart"]
# print the whole list
print(beatles)
```

### Calculate and print the length of a list

```python
# calculate number of beatles
number_of_beatles = len(beatles)
# print number of beatles
print(f"There are {number_of_beatles} members of the Beatles.")
```

### Print an item in a list 
   
   _see PowerPoint slides_

```python
# remind ourselves of the list
print(beatles)

# print the first beatle
print(beatles[0])

# print the last Beatle
print(beatles[-1])
print(beatles[4])
print(beatles[len(beatles) - 1])

# many ways to print George
print(beatles[2])
print(beatles[-3])

```

### Remove an item from a list

```python
beatles = ["John", "Paul", "George", "Ringo", "Stuart"]
# Remove the last item in the list (bye, Stuart), popping the list
# stu = beatles.pop()
beatles.pop()
# What happened to the Beatles?
# print(stu)
print(beatles)
```

### Make a new list

```python
# make a list called 'stones' with the Rolling Stones (Mick, Ron, and Keith)
stones = ["Mick", "Ron", "Keith"]
```

### Add two lists

```python
# print the two groups
print("The two groups separately: {}, {}".format(beatles, stones))

# add two lists together
supergroup = beatles + stones

# print our new list
print(f"{len(supergroup)} members in our new band: {supergroup}")
```

### Growing a list with `extend()`

```python
# print our lists
print("Jackson 5:", jackson5)
print("Destiny's child:", destiny)
# use extend() to add Destiny's Child to the Jackson 5
```

```python

# extend() modifies a list in place, it doesn't return a value (i.e., a new list). You don't add lists together and assign that to a new variable
jackson5.extend(destiny)
# here is what extend is doing:
# jackson5 = jackson5 + destiny

print(f"Here is the extended jackson5: {jackson5}")
```

### Growing a list with `append()`

```python
# use `append()` to add some old men
num = [1, 2, 3]
let = ['a', 'b', 'c']
print(f"our lists: {num, let}")

# append let onto num
num.append(let)
# What did append do?
print(f"After appending let to num: {num}")

# The last item in our list is itself a list. Append always adds a single element to the list, regardless of what it is (including lists themselves)

more_num = [4, 5, 6]
more_num.append(7)
print(more_num)
i = 8
more_num.append(i)
print(more_num)
```

```python
# How long is this list?
print(len(num))

# print the letters
print(num[-1])
print(num[3])
```

```python
# find and print 'b'
# here's how to get the list
print(type(num[3]))

# here's how to get an item in the list, 'b' for example
print(num[3][1])
```

### Strings as lists

- strings are just a list of characters

- this means you can index into a string, the same way you would a list

```python
name = 'beyonce'
print(name[0], name[-1])
```

```python
# here's a gene with structure EXONintronEXON
gene = 'ATGCCAGTttccggaaGTTCTAA'

# extract the first exon (i.e, the first string of capital letters: 'ATGCCAGT')
# [inclusive, exclusive]
print(f"The first exon: {gene[0:8]}")

# get the start codon (the first three characters of the gene)
print(gene[0:3])

# get the stop codon (the last three characters of the gene)
print(gene[-3:])
```

```python
# split our gene into a list
gene_as_list = list(gene)
print(gene_as_list)
print(gene)
```
